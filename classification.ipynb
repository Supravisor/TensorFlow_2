```python
!pip install tensorflow
!pip install pandas
!pip install numpy

import tensorflow as tf
import pandas as pd
import numpy as np

# Define names and categories
CSV_COLUMN_NAMES = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth', 'Species']
SPECIES = ['Setosa', 'Versicolor', 'Virginica']

# Download the datasets
train_path = tf.keras.utils.get_file('iris_training.csv', 'https://googleapis.com')
test_path = tf.keras.utils.get_file('iris_test.csv', 'https://googleapis.com')

# Load dataset into Pandas DataFrames
train = pd.read_csv(train_path, names=CSV_COLUMN_NAMES, header=0)
test = pd.read_csv(test_path, names=CSV_COLUMN_NAMES, header=0)

# Separate features from labels
train_y = train.pop('Species')
test_y = test.pop('Species')

# Build the modern Keras Neural Network
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),          # 4 input features
    tf.keras.layers.Dense(30, activation='relu'), # First hidden layer (30 units)
    tf.keras.layers.Dense(10, activation='relu'), # Second hidden layer (10 units)
    tf.keras.layers.Dense(3, activation='softmax') # Output layer (3 classes)
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
model.fit(
    train.values, 
    train_y.values, \
    epochs=150, 
    batch_size=256, 
    shuffle=True,
    verbose=1
)

# Evaluate the model on the test set
loss, accuracy = model.evaluate(test.values, test_y.values, verbose=0)
print(f'\nTest set accuracy: {accuracy:0.3f}\n')

# --- INTERACTIVE PREDICTION SECTION (FIXED FOR TF2/KERAS) ---
features = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth']
predict_values = []

print('Please type numeric values as prompted.')
for feature in features:
    while True:
        val = input(feature + ': ')
        try:
            predict_values.append(float(val))
            break
        except ValueError:
            print("Invalid input. Please enter a valid number.")

# Reshape list to standard Keras expected shape: (batch_size, num_features) -> (1, 4)
input_data = np.array([predict_values])

# Run Keras prediction pipeline
predictions = model.predict(input_data, verbose=0)

# Extract prediction results
probabilities = predictions[0]
class_id = np.argmax(probabilities)
probability = probabilities[class_id]

print(f'\nPrediction is {SPECIES[class_id]} ({100 * probability:.1f}%)')
```
